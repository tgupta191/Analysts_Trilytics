```python
# Fill missing values
df["Ownership"] = df["Ownership"].fillna("Nomad")
df["Address type"] = df["Address type"].fillna("Nomad")
df["Location"] = df["Location"].fillna("0,0")

# Save to new Excel
df.to_excel("TrainData_filled.xlsx", index=False)
```

```python
# Columns to exclude from imputation
exclude_cols = ['Address type', 'Ownership', 'Avg_Disbursement_Amount_Bureau', 'Location']

# Impute for numeric columns except excluded
for col in df.select_dtypes(include=['float', 'int']).columns:
    if col not in exclude_cols:
        df[col] = df[col].fillna(df[col].median())

# Impute for object (categorical) columns except excluded
for col in df.select_dtypes(include=['object']).columns:
    if col not in exclude_cols:
        df[col] = df[col].fillna(df[col].mode()[0])
# Save updated file
df.to_excel("TrainData_imputed_median.xlsx", index=False)
```
